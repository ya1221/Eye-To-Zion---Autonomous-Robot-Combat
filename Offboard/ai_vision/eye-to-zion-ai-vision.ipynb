```python
import os
from roboflow import Roboflow

api_key = os.environ.get("ROBOFLOW_API_KEY")
if not api_key:
    raise SystemExit("ROBOFLOW_API_KEY is not set (Kaggle: Add-ons -> Secrets, then attach it to this notebook).")

rf = Roboflow(api_key=api_key)
project = rf.workspace("yahavs-workspace-ystrz").project("eye-to-zion-detection-robots")
version = project.version(4)
dataset = version.download("yolo26")
```

# Model Training

```python
from ultralytics import YOLO

model = YOLO('yolo26n.pt')

results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    project='EyeToZion_AI',
    name='yolo26_robot_detect',
    hsv_h=0.015,   # Minor hue shift
    hsv_s=0.7,     # Saturation for variable lighting
    hsv_v=0.4,     # Exposure for shadows/highlights
    translate=0.1, # Image shift
    scale=0.5,     # Zoom for varying distances
    fliplr=0.5     # Horizontal flip
)
```

# Results & Download

```python
import shutil
import os
from IPython.display import FileLink

folder_to_zip = 'EyeToZion_AI/yolo26_robot_detect'
zip_name = 'yolo26_full_results'

if os.path.exists(folder_to_zip):
    shutil.make_archive(zip_name, 'zip', folder_to_zip)
    print("Download Full Training Folder (ZIP):")
    display(FileLink(f"{zip_name}.zip"))
else:
    print("Training folder not found.")
```
